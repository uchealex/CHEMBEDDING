```python
# cell0

!pip install torch transformers scikit-learn matplotlib joblib

import torch
from transformers import AutoTokenizer

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Same ChemBERTa tokenizer as in the first notebook
tokenizer = AutoTokenizer.from_pretrained('seyonec/ChemBERTa-zinc-base-v1')
```

```python
import torch
import torch.nn as nn
from sklearn.metrics import r2_score, mean_absolute_error
from transformers import AutoModel
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from torch.utils.data import Dataset, DataLoader
import joblib
from tqdm.auto import tqdm # Import tqdm for progress bars

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}") # Print the device being used

# ===== 1) Load saved artifacts from previous Colab =====

# Peptide encoder weights and scaler
peptide_bert_state = torch.load('peptide_bert_encoder.pth', map_location=device)
scaler = joblib.load('peptide_rt_scaler.pkl')

# Partial lipid metrics from 5%, 25%, 50%
partial_metrics = joblib.load('lipid_metrics_partial_5_25_50.pkl')
percentages      = partial_metrics['percentages']
r2_train_TL      = partial_metrics['r2_train_TL']
r2_test_TL       = partial_metrics['r2_test_TL']
mae_train_TL     = partial_metrics['mae_train_TL']
mae_test_TL      = partial_metrics['mae_test_TL']
r2_train_Base    = partial_metrics['r2_train_Base']
r2_test_Base     = partial_metrics['r2_test_Base']
mae_train_Base   = partial_metrics['mae_train_Base']
mae_test_Base    = partial_metrics['mae_test_Base']

# ===== 2) Rebuild lipid datasets and scaling =====

lipid_train_df = pd.read_csv('/content/METLIN_RT_lipid_train_with_rdkit.csv')
lipid_test_df  = pd.read_csv('/content/METLIN_RT_lipid_test_with_rdkit.csv')

lipid_train_smiles = lipid_train_df['smile'].tolist()
lipid_test_smiles  = lipid_test_df['smile'].tolist()

rt_train = lipid_train_df['rt'].astype(float).values
rt_test  = lipid_test_df['rt'].astype(float).values

# Get RT scaling params from loaded scaler (feature 0)
rt_scale = scaler.scale_[0]
rt_min   = scaler.min_[0]

rt_train_scaled = rt_train * rt_scale + rt_min
rt_test_scaled  = rt_test  * rt_scale + rt_min

lipid_train_df['rt_scaled'] = rt_train_scaled.astype(np.float32)
lipid_test_df['rt_scaled']  = rt_test_scaled.astype(np.float32)

def inv_minmax_rt(z):
    return (z - rt_min) / rt_scale

class LipidRTDataset(Dataset):
    def __init__(self, smiles, rt_scaled, tokenizer, max_len=128):
        self.smiles = smiles
        self.rt_scaled = rt_scaled
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.smiles)

    def __getitem__(self, idx):
        enc = tokenizer(self.smiles[idx],
                        padding='max_length',
                        truncation=True,
                        max_length=self.max_len,
                        return_tensors='pt')
        return {
            'input_ids':      enc['input_ids'].squeeze(0),
            'attention_mask': enc['attention_mask'].squeeze(0),
            'labels':         torch.tensor(self.rt_scaled[idx], dtype=torch.float32).unsqueeze(0)
        }

class ChemBERTaRTSingle(nn.Module):
    def __init__(self):
        super().__init__()
        self.bert = AutoModel.from_pretrained('seyonec/ChemBERTa-zinc-base-v1')
        hs = self.bert.config.hidden_size
        self.regressor = nn.Linear(hs, 1)

    def forward(self, input_ids, attention_mask):
        out = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        cls = out.last_hidden_state[:, 0, :]
        return self.regressor(cls)

def get_preds_and_truth_rt(loader, model_, device_):
    model_.eval()
    ys, yps = [], []
    with torch.no_grad():
        for b in loader:
            inp = b['input_ids'].to(device_)
            msk = b['attention_mask'].to(device_)
            out = model_(inp, msk).cpu().numpy()
            ys.append(b['labels'].cpu().numpy())
            yps.append(out)
    return np.vstack(ys).squeeze(1), np.vstack(yps).squeeze(1)

num_epochs = 15
batch_size_lipid = 16

print("\n===== Continuing lipid fits: 75% and 100% (new Colab) =====")

percentages_part2 = [75, 100]
fraction_values_part2 = [p / 100.0 for p in percentages_part2]

# To collect last-epoch test predictions for 100% fits
tl_100_y_true = None
tl_100_y_pred = None
base_100_y_true = None
base_100_y_pred = None

for p, frac in zip(percentages_part2, fraction_values_part2):
    print(f"\n===== Percentage: {p}% of lipid training data ====")

    n_train = max(1, int(len(lipid_train_df) * frac))
    lipid_subset = lipid_train_df.sample(n=n_train, random_state=42)

    subset_smiles = lipid_subset['smile'].tolist()
    subset_rt_scaled  = lipid_subset['rt_scaled'].values.astype(np.float32)

    train_dataset_lipid = LipidRTDataset(subset_smiles, subset_rt_scaled, tokenizer)
    test_dataset_lipid  = LipidRTDataset(
        lipid_test_smiles,
        lipid_test_df['rt_scaled'].values.astype(np.float32),
        tokenizer
    )

    train_loader_lipid = DataLoader(train_dataset_lipid, batch_size=batch_size_lipid, shuffle=True)
    test_loader_lipid  = DataLoader(test_dataset_lipid,  batch_size=batch_size_lipid, shuffle=False)

    # ---- Transfer learning model ----
    print("  -> Training TRANSFER model (peptide-initialized)")
    tl_model = ChemBERTaRTSingle().to(device)
    tl_model.bert.load_state_dict(peptide_bert_state)  # use loaded encoder

    optimizer_tl = torch.optim.AdamW(tl_model.parameters(), lr=2.5e-5)
    criterion_tl = nn.MSELoss()

    for epoch in range(1, num_epochs + 1):
        tl_model.train()
        total_loss_tl = 0.0
        # Add tqdm to the training loop
        for b in tqdm(train_loader_lipid, desc=f"TL Epoch {epoch}/{num_epochs} Train"):
            optimizer_tl.zero_grad()
            out = tl_model(b['input_ids'].to(device), b['attention_mask'].to(device))
            loss = criterion_tl(out, b['labels'].to(device))
            loss.backward()
            optimizer_tl.step()
            total_loss_tl += loss.item() * b['input_ids'].size(0)
        avg_loss_tl = total_loss_tl / len(train_dataset_lipid)

        y_tr_sc, yh_tr_sc = get_preds_and_truth_rt(train_loader_lipid, tl_model, device)
        y_te_sc, yh_te_sc = get_preds_and_truth_rt(test_loader_lipid,  tl_model, device)

        y_tr_true = inv_minmax_rt(y_tr_sc)
        y_tr_pred = inv_minmax_rt(yh_tr_sc)
        y_te_true = inv_minmax_rt(y_te_sc)
        y_te_pred = inv_minmax_rt(yh_te_sc)

        r2_tr = r2_score(y_tr_true, y_tr_pred)
        r2_te = r2_score(y_te_true, y_te_pred)
        mae_tr = mean_absolute_error(y_tr_true, y_tr_pred)
        mae_te = mean_absolute_error(y_te_true, y_te_pred)

        r2_train_TL[p].append(r2_tr)
        r2_test_TL[p].append(r2_te)
        mae_train_TL[p].append(mae_tr)
        mae_test_TL[p].append(mae_te)

        print(f"    TL Epoch {epoch}/{num_epochs} | Loss {avg_loss_tl:.4f} | "
              f"R² Train {r2_tr:.4f} Test {r2_te:.4f} | "
              f"MAE Train {mae_tr:.2f} Test {mae_te:.2f}")

        if p == 100 and epoch == num_epochs:
            tl_100_y_true = y_te_true.copy()
            tl_100_y_pred = y_te_pred.copy()

    # ---- Baseline model ----
    print("  -> Training BASELINE model (fresh ChemBERTa)")
    base_model = ChemBERTaRTSingle().to(device)
    optimizer_base = torch.optim.AdamW(base_model.parameters(), lr=2.5e-5)
    criterion_base = nn.MSELoss()

    for epoch in range(1, num_epochs + 1):
        base_model.train()
        total_loss_base = 0.0
        # Add tqdm to the training loop
        for b in tqdm(train_loader_lipid, desc=f"Base Epoch {epoch}/{num_epochs} Train"):
            optimizer_base.zero_grad()
            out = base_model(b['input_ids'].to(device), b['attention_mask'].to(device))
            loss = criterion_base(out, b['labels'].to(device))
            loss.backward()
            optimizer_base.step()
            total_loss_base += loss.item() * b['input_ids'].size(0)
        avg_loss_base = total_loss_base / len(train_dataset_lipid)

        y_tr_sc, yh_tr_sc = get_preds_and_truth_rt(train_loader_lipid, base_model, device)
        y_te_sc, yh_te_sc = get_preds_and_truth_rt(test_loader_lipid,  base_model, device)

        y_tr_true = inv_minmax_rt(y_tr_sc)
        y_tr_pred = inv_minmax_rt(yh_tr_sc)
        y_te_true = inv_minmax_rt(y_te_sc)
        y_te_pred = inv_minmax_rt(yh_te_sc)

        r2_tr = r2_score(y_tr_true, y_tr_pred)
        r2_te = r2_score(y_te_true, y_te_pred)
        mae_tr = mean_absolute_error(y_tr_true, y_tr_pred)
        mae_te = mean_absolute_error(y_te_true, y_te_pred)

        r2_train_Base[p].append(r2_tr)
        r2_test_Base[p].append(r2_te)
        mae_train_Base[p].append(mae_tr)
        mae_test_Base[p].append(mae_te)

        print(f"    Base Epoch {epoch}/{num_epochs} | Loss {avg_loss_base:.4f} | "
              f"R² Train {r2_tr:.4f} Test {r2_te:.4f} | "
              f"MAE Train {mae_tr:.2f} Test {mae_te:.2f}")

        if p == 100 and epoch == num_epochs:
            base_100_y_true = y_te_true.copy()
            base_100_y_pred = y_te_pred.copy()

print("\n✅ Finished all lipid fits (5%, 25%, 50%, 75%, 100%) in combined metrics.")

# Optionally save full metrics after completing 75% and 100%
full_metrics = {
    'percentages': percentages,
    'r2_train_TL': r2_train_TL,
    'r2_test_TL': r2_test_TL,
    'mae_train_TL': mae_train_TL,
    'mae_test_TL': mae_test_TL,
    'r2_train_Base': r2_train_Base,
    'r2_test_Base': r2_test_Base,
    'mae_train_Base': mae_train_Base,
    'mae_test_Base': mae_test_Base,
}
joblib.dump(full_metrics, 'lipid_metrics_all_5_100.pkl')
print("✅ Saved full lipid metrics to 'lipid_metrics_all_5_100.pkl'.")

# ===== 3) Compact boxplots: distinct colors per model =====

fig, (ax_r2, ax_mae) = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

categories = [f"{p}%" for p in percentages]
n_cat = len(percentages)

chunk_width = 2.5
centers = np.arange(n_cat) * chunk_width
offsets = np.array([-0.9, -0.3, 0.3, 0.9])  # TL_tr, TL_te, Base_tr, Base_te

pos_TL_tr   = centers + offsets[0]
pos_TL_te   = centers + offsets[1]
pos_Base_tr = centers + offsets[2]
pos_Base_te = centers + offsets[3]

# R² boxplots
data_TL_tr_r2   = [r2_train_TL[p]   for p in percentages]
data_TL_te_r2   = [r2_test_TL[p]    for p in percentages]
data_Base_tr_r2 = [r2_train_Base[p] for p in percentages]
data_Base_te_r2 = [r2_test_Base[p]  for p in percentages]

bp_TL_tr_r2   = ax_r2.boxplot(data_TL_tr_r2,   positions=pos_TL_tr,   widths=0.35, patch_artist=True)
bp_TL_te_r2   = ax_r2.boxplot(data_TL_te_r2,   positions=pos_TL_te,   widths=0.35, patch_artist=True)
bp_Base_tr_r2 = ax_r2.boxplot(data_Base_tr_r2, positions=pos_Base_tr, widths=0.35, patch_artist=True)
bp_Base_te_r2 = ax_r2.boxplot(data_Base_te_r2, positions=pos_Base_te, widths=0.35, patch_artist=True)

for box in bp_TL_tr_r2['boxes'] + bp_TL_te_r2['boxes']:
    box.set_facecolor('C0')
    box.set_alpha(0.7)
for box in bp_Base_tr_r2['boxes'] + bp_Base_te_r2['boxes']:
    box.set_facecolor('C1')
    box.set_alpha(0.7)

ax_r2.set_ylabel("R² (RT)")
ax_r2.set_title("R² by Lipid Training Percentage and Model Type")
ax_r2.set_ylim(0.0, 1.0)
ax_r2.set_xticks(centers)
ax_r2.set_xticklabels(categories)
ax_r2.grid(True, linestyle='--', alpha=0.3)
ax_r2.legend(
    [bp_TL_tr_r2["boxes"][0],
     bp_Base_tr_r2["boxes"][0]],
    ["Transfer (peptide-pretrained)", "Baseline (fresh ChemBERTa)"],
    loc="lower right"
)

# MAE boxplots
data_TL_tr_mae   = [mae_train_TL[p]   for p in percentages]
data_TL_te_mae   = [mae_test_TL[p]    for p in percentages]
data_Base_tr_mae = [mae_train_Base[p] for p in percentages]
data_Base_te_mae = [mae_test_Base[p]  for p in percentages]

bp_TL_tr_mae   = ax_mae.boxplot(data_TL_tr_mae,   positions=pos_TL_tr,   widths=0.35, patch_artist=True)
bp_TL_te_mae   = ax_mae.boxplot(data_TL_te_mae,   positions=pos_TL_te,   widths=0.35, patch_artist=True)
bp_Base_tr_mae = ax_mae.boxplot(data_Base_tr_mae, positions=pos_Base_tr, widths=0.35, patch_artist=True)
bp_Base_te_mae = ax_mae.boxplot(data_Base_te_mae, positions=pos_Base_te, widths=0.35, patch_artist=True)

for box in bp_TL_tr_mae['boxes'] + bp_TL_te_mae['boxes']:
    box.set_facecolor('C0')
    box.set_alpha(0.7)
for box in bp_Base_tr_mae['boxes'] + bp_Base_te_mae['boxes']:
    box.set_facecolor('C1')
    box.set_alpha(0.7)

ax_mae.set_ylabel("MAE (RT)")
ax_mae.set_title("MAE by Lipid Training Percentage and Model Type")
ax_mae.set_xticks(centers)
ax_mae.set_xticklabels(categories)
ax_mae.grid(True, linestyle='--', alpha=0.3)
ax_mae.legend(
    [bp_TL_tr_mae["boxes"][0],
     bp_Base_tr_mae["boxes"][0]],
    ["Transfer (peptide-pretrained)", "Baseline (fresh ChemBERTa)"],
    loc="upper right"
)

plt.tight_layout()
plt.show()

# ===== 4) Predicted vs Observed RT scatterplots for 100% lipid (test set) =====

if tl_100_y_true is not None and base_100_y_true is not None:
    fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharex=True, sharey=True)

    # Transfer model scatter
    ax = axes[0]
    ax.scatter(tl_100_y_true, tl_100_y_pred, alpha=0.6)
    min_rt = min(tl_100_y_true.min(), tl_100_y_pred.min())
    max_rt = max(tl_100_y_true.max(), tl_100_y_pred.max())
    ax.plot([min_rt, max_rt], [min_rt, max_rt], linestyle='--')
    ax.set_xlabel("Observed RT")
    ax.set_ylabel("Predicted RT")
    ax.set_title("Transfer model (100% lipid test)")

    # Baseline model scatter
    ax = axes[1]
    ax.scatter(base_100_y_true, base_100_y_pred, alpha=0.6)
    min_rt_b = min(base_100_y_true.min(), base_100_y_pred.min())
    max_rt_b = max(base_100_y_true.max(), base_100_y_pred.max())
    ax.plot([min_rt_b, max_rt_b], [min_rt_b, max_rt_b], linestyle='--')
    ax.set_xlabel("Observed RT")
    ax.set_title("Baseline model (100% lipid test)")

    plt.tight_layout()
    plt.show()
else:
    print("⚠️ 100% lipid predictions for TL / Base not found; ensure p==100 loop ran to completion.")
```

```python
# === cell: save boxplot + scatterplot data to CSV ===
import pandas as pd
import numpy as np

# -----------------------------
# 1) Save boxplot data (R² + MAE)
# -----------------------------
boxplot_rows = []

for p in percentages:  # e.g. [5, 25, 50, 75, 100]
    # For each model type
    for model_name, r2_tr_dict, r2_te_dict, mae_tr_dict, mae_te_dict in [
        ("Transfer", r2_train_TL, r2_test_TL, mae_train_TL, mae_test_TL),
        ("Baseline", r2_train_Base, r2_test_Base, mae_train_Base, mae_test_Base),
    ]:
        # Train / Test splits
        for split_name, r2_dict, mae_dict in [
            ("Train", r2_tr_dict, mae_tr_dict),
            ("Test",  r2_te_dict, mae_te_dict),
        ]:
            r2_values  = r2_dict[p]
            mae_values = mae_dict[p]

            # One row per epoch value
            for epoch_idx, (r2_val, mae_val) in enumerate(zip(r2_values, mae_values), start=1):
                boxplot_rows.append({
                    "percentage": p,
                    "model": model_name,       # "Transfer" or "Baseline"
                    "split": split_name,       # "Train" or "Test"
                    "epoch": epoch_idx,
                    "metric": "R2",
                    "value": r2_val,
                })
                boxplot_rows.append({
                    "percentage": p,
                    "model": model_name,
                    "split": split_name,
                    "epoch": epoch_idx,
                    "metric": "MAE",
                    "value": mae_val,
                })

boxplot_df = pd.DataFrame(boxplot_rows)
boxplot_df.to_csv("lipid_boxplot_data_long.csv", index=False)
print("✅ Saved boxplot data to 'lipid_boxplot_data_long.csv'")
print(boxplot_df.head())

# -----------------------------
# 2) Save scatterplot data (100% lipid, test set)
# -----------------------------
if tl_100_y_true is not None and base_100_y_true is not None:
    scatter_rows = []

    # Transfer model
    for obs, pred in zip(tl_100_y_true, tl_100_y_pred):
        scatter_rows.append({
            "model": "Transfer",
            "percentage": 100,
            "split": "Test",
            "observed_rt": obs,
            "predicted_rt": pred,
        })

    # Baseline model
    for obs, pred in zip(base_100_y_true, base_100_y_pred):
        scatter_rows.append({
            "model": "Baseline",
            "percentage": 100,
            "split": "Test",
            "observed_rt": obs,
            "predicted_rt": pred,
        })

    scatter_df = pd.DataFrame(scatter_rows)
    scatter_df.to_csv("lipid_scatter_100pct_test.csv", index=False)
    print("✅ Saved scatter data to 'lipid_scatter_100pct_test.csv'")
    print(scatter_df.head())
else:
    print("⚠️ Cannot save scatter data: tl_100_y_true/base_100_y_true are None. "
          "Make sure the 100% runs completed in the previous cell.")
```
